**4. Describe how dictionaries store data.**

Dictionaries in Python store data as key-value pairs. Here's how it works:

- Keys: Unique identifiers (strings, integers, tuples) that map to specific values.
- Values: The data associated with each key (can be any data type).
- Hash table: Dictionaries use a hash table internally to store key-value pairs, allowing for fast lookups.

When you add a key-value pair:

1. Python hashes the key to generate an index.
2. The value is stored at that index in the hash table.

When you look up a key:

1. Python hashes the key to generate the index.
2. It retrieves the value stored at that index.

This makes dictionaries super fast for lookups, insertions, and deletions – typically O(1) time complexity.


my_dict = {'name': 'John', 'age': 30}

print(my_dict['name'])  # Output: John

**5. Why might you use a set instead of a list in Python?**

You might use a set instead of a list in Python when:

- **You need unique elements:** Sets automatically remove duplicates, ensuring all elements are unique.
- **You need fast membership testing:** Checking if an element is in a set is typically faster (O(1)) than in a list (O(n)).
- **You need set operations:** Sets support operations like union, intersection, and difference.

However, sets are unordered and don't support indexing or slicing, so use them when order doesn't matter and you need fast lookups.

 **8. What is a hash table, and how does it relate to dictionaries in Python?**

 A hash table is a data structure that stores key-value pairs in an array using a hash function to map keys to indices.

In Python, dictionaries (dict) are implemented as hash tables. Here's how it works:

- Key hashing: When you add a key-value pair, Python hashes the key to generate an index.
- Storage: The value is stored at that index in the hash table.
- Lookup: When you access a key, Python hashes it and looks up the value at the corresponding index.

This makes dictionaries super fast for lookups, insertions, and deletions – typically O(1) time complexity.


my_dict = {'name': 'John', 'age': 30}

**10.  Explain why strings are immutable in Python.**

Strings in Python are immutable because of a combination of design choices and performance considerations:

**14.  How does the “in” keyword work differently for lists and dictionaries?**

The in keyword works differently for lists and dictionaries:

- Lists: For lists, in checks if an element is present by iterating through the list (O(n) time complexity).
- Dictionaries: For dictionaries, in checks if a key is present using the hash table (O(1) average time complexity).


**List example**

my_list = [1, 2, 3]

print(2 in my_list)  # Output: True (O(n) lookup)

**Dictionary example**

my_dict = {'a': 1, 'b': 2}

print('a' in my_dict)  # Output: True (O(1) lookup)

**17. Describe the time complexity of accessing elements in a dictionary.**

The time complexity of accessing elements in a dictionary is typically O(1), making it very efficient.

Here's why:

- Hashing: Python dictionaries use a hash table to store key-value pairs.
- Index calculation: When you access a key, Python calculates the index using the hash function.
- Collision resolution: In case of collisions (multiple keys hashing to the same index), Python uses techniques like open addressing or chaining.

While the worst-case scenario can be O(n), it's extremely rare and usually occurs when the hash function is poorly designed or the dictionary is heavily loaded.

In general, dictionary lookups are considered O(1) operations, making them ideal for fast data retrieval.

**18. In what situations are lists preferred over dictionaries?**

Lists are preferred over dictionaries in situations where:

Examples include:

**19. Why are dictionaries considered unordered, and how does that affect data retrieval**

Dictionaries in Python are considered unordered because they store key-value pairs in a hash table, which doesn't preserve any particular order. Before Python 3.7, the iteration order was unpredictable.

However, in Python 3.7 and later:

- Order is preserved: Dictionaries maintain their insertion order.
- Fast lookups: You can still access values using their keys (O(1) average time complexity).


my_dict = {'a': 1, 'b': 2, 'c': 3}

print(my_dict['b'])  # Output: 2 (fast lookup)

**Iteration order is preserved**

for key in my_dict:

    print(key)  # Output: a, b, c

**20.  Explain the difference between a list and a dictionary in terms of data retrieval.**

Lists and dictionaries differ in data retrieval:

- Lists: Retrieve data by index (position). my_list[0] gets the first element. Time complexity: O(1) for index-based access, O(n) for searching.
- Dictionaries: Retrieve data by key. my_dict['key'] gets the associated value. Time complexity: O(1) average for key-based access.


**List example**

my_list = [1, 2, 3]

print(my_list[0])  # Output: 1 (index-based access)

**Dictionary example**

my_dict = {'a': 1, 'b': 2}

print(my_dict['a'])  # Output: 1 (key-based access)
